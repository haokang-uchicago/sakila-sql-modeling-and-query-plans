### 3.2b Claude's answer (based on the three EXPLAIN outputs)

**Prompt:** For each of these three queries and their EXPLAIN output: for every row of the plan, tell me the table, the access type (ALL, index, range, ref, eq_ref, const), the key used and the estimated rows — and which table is read in full, and why.

**Plan 1 — Q1 (single-table filter):** `SELECT title, description FROM sakila.film WHERE rating = 'PG-13'`

| table | type | key | rows | notes |
|---|---|---|---|---|
| film | ALL | none | 1000 | filtered = 20.0; Extra: Using where |

- **Read in full:** `film`. `possible_keys` is None because there is no index on `rating`, so MySQL reads all 1000 rows and tests each one against the WHERE condition ("Using where").
- `filtered` = 20% is the optimizer's guess that about 200 of 1000 rows match (no index statistics on rating).

**Plan 2 — Q17 (GROUP BY aggregate):** `SELECT rating, count(*) FROM sakila.film WHERE rating IN (...) GROUP BY rating`

| table | type | key | rows | notes |
|---|---|---|---|---|
| film | ALL | none | 1000 | filtered = 50.0; Extra: Using where; Using temporary |

- **Read in full:** `film`, for the same reason: no index on `rating`.
- **"Using temporary":** with no index to read rows in rating order, MySQL builds a temporary table to collect and count the groups.
- **The `filtered` estimate is wrong:** the optimizer expects 50% of rows to match, but every film has one of the five listed ratings, so the true figure is 100%.

**Plan 3 — Q26 (6-table join)**

| # | table | type | key | ref | rows |
|---|---|---|---|---|---|
| 0 | cat (category) | ALL | none (PRIMARY possible, not used) | — | 16 |
| 1 | fc (film_category) | ref | fk_film_category_category | cat.category_id | 62 |
| 2 | f (film) | eq_ref | PRIMARY | fc.film_id | 1 |
| 3 | i (inventory) | ref | idx_fk_film_id | fc.film_id | 4 |
| 4 | r (rental) | ref | idx_fk_inventory_id | i.inventory_id | 3 |
| 5 | c (customer) | eq_ref | PRIMARY | r.customer_id | 1 |

- **Join order:** MySQL does not follow the written order. It starts from `category` (smallest table, and the one with the filter `name = 'Horror'`) and walks outward through indexes: category → film_category → film → inventory → rental → customer.
- **Read in full:** only `category` (type ALL). There is no index on `name`, but the table has only 16 rows, so a full scan is cheap.
- **Access types:** `eq_ref` (film, customer) = primary-key lookup, at most 1 row per lookup; `ref` (film_category, inventory, rental) = non-unique index, several rows can match.
- **The largest table, `rental`** (about 16,000 rows), is **not** scanned; it is reached through `idx_fk_inventory_id` at about 3 rows per inventory copy.
- **Rough work estimate:** 16 × 62 × 1 × 4 × 3 × 1 ≈ 12,000 row combinations.
- **Caveat from Claude:** the `filtered` and `Extra` columns for Plan 3 were cut off on screen; Claude guessed "Using where" on `cat` and "Using temporary" for the GROUP BY.
